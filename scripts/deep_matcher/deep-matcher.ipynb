```python
import pandas as pd

files = ["train", "valid", "test"]

for file in files:
    df = pd.read_excel(f"Deep Matcher/{file}"+".xlsx")
    df.to_csv(f"Deep Matcher/{file}"+".csv", index=False)
```

```python
import deepmatcher as dm

train, validation, test = dm.data.process(
    path='Deep Matcher',
    train='train.csv',
    validation='valid.csv',
    test='test.csv',
    ignore_columns=('left_row_index', 'right_row_index'),
    left_prefix='left_',
    right_prefix='right_')
```

```python
model = dm.MatchingModel(attr_summarizer='hybrid')
```

```python
model.run_train(
    train,
    validation,
    epochs=10,
    batch_size=12,
    best_save_path='hybrid_model.pth',
    pos_neg_ratio=2)
```

```python
model.run_eval(test)
```

```python
import pandas as pd

blocking = ["company_blocking", "industry_blocking", "socialmedia_blocking", "finance_blocking"]

for file in blocking:
    df = pd.read_excel(f"blocking_excels/{file}"+".xlsx")
    df.to_csv(f"blocking_excels/{file}"+".csv", index=False)
```

### Company Predictions

```python
unlabeled = dm.data.process_unlabeled(
    path='blocking_excels/company_blocking.csv',
    trained_model=model)
```

```python
predictions = model.run_prediction(unlabeled, output_attributes=True)
```

```python
predictions.to_csv('blocking_excels/company_predictions.csv')
```
