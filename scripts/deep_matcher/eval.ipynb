```python
import pandas as pd
from sklearn.metrics import f1_score, accuracy_score

pred_folder = "../../Deep Matcher/Predictions/"
gt_folder = "../../record_linkage_groundtruth/"
prefix_predictions_file = '_predictions.csv'
prefix_groundtruth_file = '_RL_groundtruth.xlsx'

files = ["company", "finance", "industry", "socialmedia"]

print("___________________________________________________\n")
for file in files:
    print(f"Evaluating Deep Matcher on {file} Predictions")
    pred_df = pd.read_csv(pred_folder+file+prefix_predictions_file)
    gt_df = pd.read_excel(gt_folder+file+prefix_groundtruth_file)
    
    merged_df = pd.merge(pred_df, gt_df, left_on='id', right_on='row_index_1')
    
    if file == "socialmedia":
        merged_df = pd.merge(pred_df, gt_df, left_on='left_name_company', right_on='name_company_1')
        
    threshold = 0.98
    merged_df['predicted_label'] = merged_df['match_score'].apply(lambda x: 1 if x >= threshold else 0)
    
    
    # True Positives (TP): Cases where both "is_match" and "predicted_label" are 1.
    # False Positives (FP): Cases where "predicted_label" is 1 but "is_match" is 0.
    # False Negatives (FN): Cases where "predicted_label" is 0 but "is_match" is 1.
    # True Negatives (TN): Cases where both are 0.
    accuracy = accuracy_score(merged_df['is_match'], merged_df['predicted_label'])
    f1 = f1_score(merged_df['is_match'], merged_df['predicted_label'], zero_division=0)
    
    print("Accuracy:", accuracy)
    print("F1 Score:", f1)
    print("___________________________________________________\n")
```
