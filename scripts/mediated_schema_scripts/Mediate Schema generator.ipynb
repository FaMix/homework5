```python
import os
import uuid
import pandas as pd

datasets_folder = '../datasets'
ownership_columns = ["ownership"]    
name_columns = ["brand name", "company", "name"]
foundedyear_columns = ["foundation year", "founded", "company_creation_date", "est_of_ownership"]
website_columns = ["url", "website", "link", "company_website"]
employees_columns = ["size", "employees", "number_of_employees"]
ceo_columns = ["ceo"]
telephone_columns = ["telephone"]
status_columns = ["company_status"]
company_columns = [ownership_columns, foundedyear_columns, website_columns, employees_columns, ceo_columns, telephone_columns, status_columns]
actual_company_columns = ["Ownership", "FoundedYear", "Website", "Employees", "CEO", "Telephone", "Status"]

# It returns a DataFrame containing the columns CompanyID, Name and Source
def name_and_id_creation(datasets_folder):
    output_df = pd.DataFrame(columns=["Name", "Source"])
    
    print('_______________ Starting initial creation of Company Dataframe with ID, Name and Source fields __________________')
    
    for filename in os.listdir(datasets_folder):
        try:
            print(f'Processing file: {filename}') 
            df = pd.read_excel(os.path.join(datasets_folder, filename))
            matching_column = extract_matching_column(name_columns, df)
            df.rename(columns={matching_column: "Name"}, inplace=True)
            df = df.drop_duplicates(subset=['Name'])
            
            df['Source'] = filename
            
            output_df = pd.concat([output_df, df[['Name', 'Source']]], ignore_index=True)
        except Exception as e:
            print(f"Error processing {filename}: {e}")
          
    output_df.insert(0, 'CompanyID', [str(uuid.uuid4()) for _ in range(len(output_df))])
    
    print('________________ Finished creating initial Company Dataframe _______________________\n\n')
    
    return output_df

def extract_matching_column(possible_columns, df):
    columns = df.columns
    try:
        matching_column = [col for col in columns if col.lower() in possible_columns][0]
    except:
        return None
    return matching_column

def extract_all_matching_columns_for_company(df):
    matching_columns = []
    for column in df.columns:
        if column in actual_company_columns:
            matching_columns.append(column)
    return matching_columns

def rename_matching_columns(df):
    for column in df.columns:
        if column.lower() in website_columns:
            df.rename(columns={column: 'Website'}, inplace=True)
            continue
        if column.lower() in ownership_columns:
            df.rename(columns={column: 'Ownership'}, inplace=True)
            continue
        if column.lower() in foundedyear_columns:
            df.rename(columns={column: 'FoundedYear'}, inplace=True)
            continue
        if column.lower() in ceo_columns:
            df.rename(columns={column: 'CEO'}, inplace=True)
            continue
        if column.lower() in status_columns:
            df.rename(columns={column: 'Status'}, inplace=True)
            continue
        if column.lower() in employees_columns:
            df.rename(columns={column: 'Employees'}, inplace=True)
            continue
        if column.lower() in telephone_columns:
            df.rename(columns={column: 'Telephone'}, inplace=True)
            continue
```

```python
def extract_company_dataframe(datasets_folder, name_and_id_df):
    output_df = name_and_id_df.copy().set_index(['Name', 'Source'])

    print('_______________ Starting to add all the data to the Initial Company Dataframe __________________')
    for filename in os.listdir(datasets_folder):
        try:
            print(f'Processing file: {filename}') 
            df = pd.read_excel(os.path.join(datasets_folder, filename))
            df['Source'] = filename
            matching_name = extract_matching_column(name_columns, df)
            df.rename(columns={matching_name: 'Name'}, inplace=True)
            rename_matching_columns(df)
            df = df.drop_duplicates(subset=['Name'])
            matching_columns = extract_all_matching_columns_for_company(df)

            new_data = df[['Name', 'Source'] + matching_columns].set_index(['Name', 'Source'])
            
            for col in new_data.columns:
                if col not in output_df.columns:
                    output_df[col] = pd.NA
            
            output_df.update(new_data)
            
        except Exception as e:
            print(f"Error processing {filename}: {e}")
    output_df = output_df.reset_index()
    desired_order = ["CompanyID", "Name", "Source", "Website", "CEO", "Ownership", "FoundedYear", "Employees", "Telephone", "Status"]
    output_df = output_df[[col for col in desired_order if col in output_df.columns]]

    print('_______________ Finished adding all data for the Company Dataframe __________________')
    
    return output_df.drop('Source', axis=1)

# TODO
# Change it in a way that "name_and_id_dataframe" becomes a global variable since it'll be used for all datasets
# 
def create_company_excel(datasets_folder):
    name_and_id_dataframe = name_and_id_creation(datasets_folder)
    df = extract_company_dataframe(datasets_folder, name_and_id_dataframe)
    df.to_excel('../Company Database Mediate.xlsx', index=False)

create_company_excel(datasets_folder)
```
